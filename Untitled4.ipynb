```python

import numpy as np
import matplotlib.pyplot as plt
from scipy.special import erfc
import os

rng = np.random.default_rng(1)

N = 100000
snr_db_list = np.array([0, 5, 10, 15, 20])
rep = 200

os.makedirs("figures", exist_ok=True)

def Q(x):
    return 0.5 * erfc(x / np.sqrt(2))


bits = rng.integers(0, 2, N)
x = 2 * bits - 1

h = (rng.standard_normal(N) + 1j * rng.standard_normal(N)) / np.sqrt(2)

ber_sim = np.zeros(len(snr_db_list))

for k, snr_db in enumerate(snr_db_list):

    snr = 10 ** (snr_db / 10)
    noise_var = 1 / snr

    n = np.sqrt(noise_var / 2) * (
        rng.standard_normal(N) + 1j * rng.standard_normal(N)
    )

    y = h * x + n

    y_eq = y / h
    bits_hat = (y_eq.real > 0).astype(int)

    err = bits_hat != bits
    ber_sim[k] = np.mean(err)

    deep = np.abs(h) ** 2 < 0.1

    print(
        f"SNR {snr_db:2d} dB: BER={ber_sim[k]:.5f} | "
        f"Deep fades={deep.mean()*100:.1f}% | "
        f"BER in fades={err[deep].mean():.3f} | "
        f"BER outside={err[~deep].mean():.5f}"
    )

    if snr_db == 10:
        y10 = y.copy()
        yeq10 = y_eq.copy()


snr_lin = 10 ** (snr_db_list / 10)

ber_rayleigh = 0.5 * (
    1 - np.sqrt(snr_lin / (1 + snr_lin))
)

ber_awgn = Q(np.sqrt(2 * snr_lin))

print("\nSNR (dB) | Sim BER | Rayleigh Theory | AWGN Theory")
print("-" * 55)

for s, a, b, c in zip(
    snr_db_list, ber_sim, ber_rayleigh, ber_awgn
):
    print(f"{s:8d} | {a:.5f} | {b:.5f} | {c:.2e}")


# 1. Transmitted and received signals

fig, ax = plt.subplots(1, 3, figsize=(15, 4))

ax[0].stem(np.arange(50), x[:50])
ax[0].set(
    title="Transmitted BPSK Symbols",
    xlabel="Sample Index",
    ylabel="Amplitude"
)

ax[1].scatter(y10.real[:2000], y10.imag[:2000], s=4, alpha=0.5)
ax[1].set(
    title="Received Signal (10 dB)",
    xlabel="In-phase",
    ylabel="Quadrature"
)

ax[2].scatter(yeq10.real[:2000], yeq10.imag[:2000], s=4, alpha=0.5)
ax[2].set(
    title="After Equalization",
    xlabel="In-phase",
    ylabel="Quadrature",
    xlim=(-5, 5),
    ylim=(-5, 5)
)

for a in ax:
    a.grid(True)

plt.tight_layout()
plt.savefig("figures/01_signal_plots.png", dpi=300, bbox_inches="tight")


# 2. Rayleigh fading envelope

r = np.linspace(0, 4, 400)

plt.figure(figsize=(7, 5))
plt.hist(np.abs(h), bins=80, density=True, alpha=0.6, label="Simulated")
plt.plot(r, 2 * r * np.exp(-r**2), "r", lw=2, label="Rayleigh PDF")

plt.xlabel("Fading Envelope |h|")
plt.ylabel("Probability Density")
plt.title("Rayleigh Fading Envelope")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.savefig("figures/02_rayleigh_envelope.png", dpi=300, bbox_inches="tight")


# 3. Deep fading

plt.figure(figsize=(9, 4))
plt.plot(20 * np.log10(np.maximum(np.abs(h[:rep]), 1e-12)))
plt.axhline(-10, color="r", linestyle="--", label="-10 dB")

plt.xlabel("Sample Index")
plt.ylabel("Channel Gain (dB)")
plt.title("Deep Fading")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.savefig("figures/03_deep_fading.png", dpi=300, bbox_inches="tight")


# 4. BER vs SNR

plt.figure(figsize=(7, 5))

plt.semilogy(snr_db_list, ber_sim, "o-", label="Simulated Rayleigh")
plt.semilogy(snr_db_list, ber_rayleigh, "k--", label="Rayleigh Theory")
plt.semilogy(snr_db_list, ber_awgn, "s:", label="AWGN Theory")

plt.xlabel("Average SNR (dB)")
plt.ylabel("Bit Error Rate (BER)")
plt.title("BPSK BER vs SNR")
plt.grid(True, which="both")
plt.legend()
plt.tight_layout()
plt.savefig("figures/04_ber_vs_snr.png", dpi=300, bbox_inches="tight")

plt.show()

print("\nAll figures saved in the 'figures' folder.")
```

```python

import numpy as np
import matplotlib.pyplot as plt
from scipy.special import erfc
import os

rng = np.random.default_rng(1)

N = 1000000
snr_db = np.array([0, 5, 10, 15, 20])
snr_lin = 10 ** (snr_db / 10)

os.makedirs("figures", exist_ok=True)

def Q(x):
    return 0.5 * erfc(x / np.sqrt(2))


# 1. Generate BPSK symbols

bits = rng.integers(0, 2, N)
x = 2 * bits - 1

ber_awgn_sim = []
ber_rayleigh_sim = []

# 2. Simulate AWGN and Rayleigh channels

for snr in snr_lin:

    noise_std = np.sqrt(1 / (2 * snr))

    # AWGN channel
    noise_awgn = noise_std * rng.standard_normal(N)
    y_awgn = x + noise_awgn

    bits_awgn = (y_awgn > 0).astype(int)
    ber_awgn_sim.append(np.mean(bits_awgn != bits))

    # Rayleigh fading channel
    h = (rng.standard_normal(N) + 1j * rng.standard_normal(N)) / np.sqrt(2)

    noise = noise_std * (
        rng.standard_normal(N) + 1j * rng.standard_normal(N)
    )

    y_rayleigh = h * x + noise

    # Channel equalization
    y_eq = y_rayleigh / h

    bits_rayleigh = (y_eq.real > 0).astype(int)
    ber_rayleigh_sim.append(np.mean(bits_rayleigh != bits))


ber_awgn_sim = np.array(ber_awgn_sim)
ber_rayleigh_sim = np.array(ber_rayleigh_sim)


# 3. Theoretical BER

ber_awgn_theory = Q(np.sqrt(2 * snr_lin))

ber_rayleigh_theory = 0.5 * (
    1 - np.sqrt(snr_lin / (1 + snr_lin))
)


# 4. Calculate fading penalty

penalty = ber_rayleigh_theory / ber_awgn_theory
ber_difference = ber_rayleigh_theory - ber_awgn_theory

print("\nBER Comparison: AWGN vs Rayleigh")
print("-" * 85)
print("SNR | AWGN Sim | Rayleigh Sim | AWGN Theory | Rayleigh Theory | Penalty")

for i in range(len(snr_db)):
    print(
        f"{snr_db[i]:3d} | "
        f"{ber_awgn_sim[i]:.6e} | "
        f"{ber_rayleigh_sim[i]:.6e} | "
        f"{ber_awgn_theory[i]:.6e} | "
        f"{ber_rayleigh_theory[i]:.6e} | "
        f"{penalty[i]:.2f}x"
    )


# 5. Identify maximum fading penalty

max_index = np.argmax(penalty)

print("\nMaximum relative fading penalty:")
print(f"SNR = {snr_db[max_index]} dB")
print(f"BER penalty = {penalty[max_index]:.2f} times")

print("\nLargest absolute BER difference:")
diff_index = np.argmax(ber_difference)
print(f"SNR = {snr_db[diff_index]} dB")
print(f"BER difference = {ber_difference[diff_index]:.6e}")


# 6. BER vs SNR comparison

plt.figure(figsize=(8, 5))

plt.semilogy(
    snr_db, ber_awgn_theory, "bo-",
    label="AWGN Theory"
)

plt.semilogy(
    snr_db, ber_rayleigh_theory, "rs-",
    label="Rayleigh Theory"
)

plt.semilogy(
    snr_db, ber_awgn_sim, "b^",
    label="AWGN Simulation"
)

plt.semilogy(
    snr_db, ber_rayleigh_sim, "rv",
    label="Rayleigh Simulation"
)

plt.xlabel("SNR (dB)")
plt.ylabel("Bit Error Rate (BER)")
plt.title("BER Comparison: AWGN vs Rayleigh Fading")
plt.grid(True, which="both")
plt.legend()
plt.tight_layout()

plt.savefig(
    "figures/awgn_vs_rayleigh_ber.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
